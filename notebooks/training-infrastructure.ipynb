```python
%%writefile /kaggle/working/Deep-Learning/src/training/engine.py

import torch

from src.training.tracker import (
    log_metrics,
    log_parameters,
    log_artifact,
)
from src.training.checkpoint import (
    save_checkpoint,
    load_checkpoint,
)


def train_one_epoch(model, train_loader, criterion, optimizer, device):
    model.train()
    train_loss_sum = 0.0
    train_correct = 0
    train_total = 0

    for images, labels in train_loader:
        images = images.to(device, non_blocking=(device.type == "cuda"))
        labels = labels.to(device, non_blocking=(device.type == "cuda"))

        optimizer.zero_grad()
        outputs = model(images)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()

        train_loss_sum += loss.item() * images.size(0)
        train_correct += (outputs.argmax(dim=1) == labels).sum().item()
        train_total += labels.size(0)

    if train_total == 0:
        raise ValueError("The training loader is empty.")

    return (
        train_loss_sum / train_total,
        100.0 * train_correct / train_total,
    )


def validate(model, val_loader, criterion, device):
    model.eval()
    val_loss_sum = 0.0
    val_correct = 0
    val_total = 0

    with torch.no_grad():
        for images, labels in val_loader:
            images = images.to(device, non_blocking=(device.type == "cuda"))
            labels = labels.to(device, non_blocking=(device.type == "cuda"))

            outputs = model(images)
            loss = criterion(outputs, labels)

            val_loss_sum += loss.item() * images.size(0)
            val_correct += (outputs.argmax(dim=1) == labels).sum().item()
            val_total += labels.size(0)

    if val_total == 0:
        raise ValueError("The validation loader is empty.")

    return (
        val_loss_sum / val_total,
        100.0 * val_correct / val_total,
    )


def fit(
    model,
    train_loader,
    val_loader,
    criterion,
    optimizer,
    device,
    epochs,
    checkpoint_path=None,
    scheduler=None,
    resume_from=None,
    last_checkpoint_path=None,
):
    history = []
    best_val_accuracy = -1.0
    start_epoch = 0

    if resume_from is not None:
        print(f"Resuming from {resume_from}")

        state = load_checkpoint(
            resume_from,
            model,
            optimizer,
            device,
            scheduler=scheduler,
            return_state=True,
        )

        start_epoch = state["epoch"]
        best_val_accuracy = state.get("best_val_accuracy", -1.0)
        history = state.get("history", [])

        print(f"Resuming at epoch {start_epoch + 1}")

    if start_epoch >= epochs:
        print("Requested epoch count already reached.")
        return history

    log_parameters({
        "epochs": epochs,
        "learning_rate": optimizer.param_groups[0]["lr"],
        "batch_size": train_loader.batch_size,
        "optimizer": optimizer.__class__.__name__,
        "loss_function": criterion.__class__.__name__,
        "resumed_from_epoch": start_epoch,
    })

    for epoch in range(start_epoch, epochs):
        print(f"\nEpoch {epoch + 1}/{epochs}")
        print("-" * 40)

        train_loss, train_accuracy = train_one_epoch(
            model, train_loader, criterion, optimizer, device
        )

        val_loss, val_accuracy = validate(
            model, val_loader, criterion, device
        )

        if scheduler is not None:
            scheduler.step(val_loss)

        current_lr = optimizer.param_groups[0]["lr"]

        epoch_results = {
            "epoch": epoch + 1,
            "learning_rate": current_lr,
            "train_loss": train_loss,
            "train_accuracy": train_accuracy,
            "val_loss": val_loss,
            "val_accuracy": val_accuracy,
        }

        history.append(epoch_results)

        log_metrics({
            "train_loss": train_loss,
            "train_accuracy": train_accuracy,
            "val_loss": val_loss,
            "val_accuracy": val_accuracy,
            "learning_rate": current_lr,
        }, step=epoch + 1)

        is_best = val_accuracy > best_val_accuracy

        if is_best:
            best_val_accuracy = val_accuracy

        if last_checkpoint_path is not None:
            save_checkpoint(
                last_checkpoint_path,
                model,
                optimizer,
                epoch + 1,
                epoch_results,
                scheduler=scheduler,
                best_val_accuracy=best_val_accuracy,
                history=history,
            )

        if checkpoint_path is not None and is_best:
            save_checkpoint(
                checkpoint_path,
                model,
                optimizer,
                epoch + 1,
                epoch_results,
                scheduler=scheduler,
                best_val_accuracy=best_val_accuracy,
                history=history,
            )
            print(f"Best checkpoint saved: {val_accuracy:.2f}%")

        print(
            f"Train Loss: {train_loss:.4f} | "
            f"Train Accuracy: {train_accuracy:.2f}%"
        )
        print(
            f"Val Loss: {val_loss:.4f} | "
            f"Val Accuracy: {val_accuracy:.2f}%"
        )

    if checkpoint_path is not None:
        log_artifact(checkpoint_path)

    if last_checkpoint_path is not None:
        log_artifact(last_checkpoint_path)

    return history
```

```python
import torch
from collections import Counter
from pathlib import Path

from src.models.scratch_cnn import ScratchCNN
from src.data.loaders import create_classification_loaders

# 1. Load the shared data
train_loader, val_loader, class_names = create_classification_loaders(
    batch_size=8,
    seed=42
)

# 2. Check class count and dataset sizes
print("===== DATASET CHECK =====")
print("Number of classes:", len(class_names))
print("First 10 classes:", class_names[:10])
print("Training images:", len(train_loader.dataset))
print("Validation images:", len(val_loader.dataset))

# 3. Check labels
train_labels = [
    int(train_loader.dataset[i][1])
    for i in range(len(train_loader.dataset))
]
val_labels = [
    int(val_loader.dataset[i][1])
    for i in range(len(val_loader.dataset))
]

print("\n===== LABEL CHECK =====")
print("Training label range:", min(train_labels), "-", max(train_labels))
print("Validation label range:", min(val_labels), "-", max(val_labels))
print("Training classes represented:", len(set(train_labels)))
print("Validation classes represented:", len(set(val_labels)))
print("Training class counts (first 10):", Counter(train_labels).most_common(10))

# 4. Check model output size
model = ScratchCNN(num_classes=len(class_names))
print("\n===== MODEL CHECK =====")
print("Model output classes:", model.classifier[-1].out_features)

# 5. Check batch dimensions and model output
images, labels = next(iter(val_loader))
with torch.no_grad():
    logits = model(images)

print("\n===== BATCH CHECK =====")
print("Images:", images.shape)
print("Labels:", labels.shape)
print("Model predictions:", logits.shape)
print("All logits finite:", torch.isfinite(logits).all().item())

# 6. Inspect saved checkpoint
checkpoint_path = Path("outputs/final_training/best_model.pth")
print("\n===== CHECKPOINT CHECK =====")
print("Checkpoint exists:", checkpoint_path.exists())

if checkpoint_path.exists():
    checkpoint = torch.load(
        checkpoint_path,
        map_location="cpu",
        weights_only=False
    )
    print("Checkpoint epoch:", checkpoint["epoch"])
    print("Checkpoint best accuracy:", checkpoint["best_val_accuracy"])
    print("Checkpoint final-layer weights:",
          checkpoint["model_state_dict"]["classifier.3.weight"].shape)
```
